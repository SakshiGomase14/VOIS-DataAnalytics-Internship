```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv('1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv')
```

```python
print("=" * 70)
print("Car Market Trends Analysis with Car Dekho Data")
print("=" * 70)

print("\nFirst 5 records:")
print(data.head())

print("\nDataset Shape:")
print(data.shape)

print("\nColumn Names:")
print(data.columns.tolist())
```

```python
print("\n" + "=" * 70)
print("BASIC INFORMATION")
print("=" * 70)

print("\nNumber of Rows:", data.shape[0])
print("Number of Columns:", data.shape[1])

print("\nData Types:")
print(data.dtypes)

print("\nMissing Values:")
print(data.isnull().sum())
```

```python
data["Depreciation"] = (
    data["Present_Price"] - data["Selling_Price"]
)

data["Depreciation_Pct"] = (
    data["Depreciation"] / data["Present_Price"]
) * 100

# Vehicle age relative to newest manufacturing year in dataset
data["Age"] = data["Year"].max() - data["Year"]

# Price retention percentage
data["Retention_Pct"] = (
    data["Selling_Price"] / data["Present_Price"]
) * 100
```

```python
print("\n Manufacturing year range")
print("Minimum Year:", data["Year"].min())
print("Maximum Year:", data["Year"].max())
```

```python
lowest_price = data["Selling_Price"].min()

print("\n Lowest Selling Price")
print("Price:", lowest_price, "lakh")

print(
    data.loc[
        data["Selling_Price"].idxmin(),
        ["Car_Name", "Year", "Selling_Price"]
    ]
)
```

```python
highest_price = data["Selling_Price"].max()

print("\n Highest Selling Price")
print("Price:", highest_price, "lakh")

print(
    data.loc[
        data["Selling_Price"].idxmax(),
        ["Car_Name", "Year", "Selling_Price"]
    ]
)
```

```python
print("\n Number of Records")
print("Total Records:", len(data))
```

```python
print("\n Missing Values")

total_missing = data.isnull().sum().sum()

if total_missing == 0:
    print("No missing values are present.")
else:
    print("Total missing values:", total_missing)

print(data.isnull().sum())
```

```python
most_depreciated = data.loc[
    data["Depreciation"].idxmax()
]

least_depreciated = data.loc[
    data["Depreciation"].idxmin()
]

print("\n Depreciation Analysis")

print("\nMost Depreciated Vehicle:")
print("Vehicle:", most_depreciated["Car_Name"])
print("Depreciation:",
      round(most_depreciated["Depreciation"], 2), "lakh")
print("Depreciation %:",
      round(most_depreciated["Depreciation_Pct"], 2), "%")

print("\nLeast Depreciated Vehicle:")
print("Vehicle:", least_depreciated["Car_Name"])
print("Depreciation:",
      round(least_depreciated["Depreciation"], 2), "lakh")
print("Depreciation %:",
      round(least_depreciated["Depreciation_Pct"], 2), "%")
```

```python
def get_brand(name):

    name = name.strip().lower()

    # Maruti
    maruti = {
        "ritz", "sx4", "ciaz", "wagon r", "swift",
        "vitara brezza", "s cross", "alto 800",
        "ertiga", "dzire", "alto k10", "ignis",
        "800", "baleno", "omni"
    }

    # Toyota
    toyota = {
        "fortuner", "innova", "corolla altis",
        "etios cross", "etios g", "etios liva",
        "corolla", "etios gd", "camry", "land cruiser"
    }

    # Hyundai
    hyundai = {
        "i20", "grand i10", "i10", "eon",
        "xcent", "elantra", "creta", "verna"
    }

    # Honda cars
    honda_cars = {
        "city", "brio", "amaze", "jazz"
    }

    if name in maruti:
        return "Maruti"

    if name in toyota:
        return "Toyota"

    if name in hyundai:
        return "Hyundai"

    if name in honda_cars:
        return "Honda"

    # Two-wheeler brands
    if "royal enfield" in name:
        return "Royal Enfield"

    if "bajaj" in name:
        return "Bajaj"

    if "ktm" in name:
        return "KTM"

    if "honda" in name or "activa" in name:
        return "Honda"

    if "hero" in name:
        return "Hero"

    if "yamaha" in name:
        return "Yamaha"

    if "tvs" in name:
        return "TVS"

    if "um " in name:
        return "UM"

    if "hyosung" in name:
        return "Hyosung"

    if "mahindra" in name:
        return "Mahindra"

    if "suzuki" in name:
        return "Suzuki"

    return "Other"


data["Brand"] = data["Car_Name"].apply(get_brand)

brand_analysis = data.groupby("Brand").agg(
    Number_of_Vehicles=("Car_Name", "size"),
    Average_Depreciation=("Depreciation", "mean"),
    Average_Depreciation_Percentage=("Depreciation_Pct", "mean")
)

brand_analysis = brand_analysis.sort_values(
    "Average_Depreciation_Percentage"
)

print("\n Brand Depreciation Analysis")
print(brand_analysis.round(2))

print("\nBrands with at least 5 records:")
print(
    brand_analysis[
        brand_analysis["Number_of_Vehicles"] >= 5
    ].round(2)
)
```

```python
print("\n Factors Affecting Depreciation")

correlation = data[
    [
        "Age",
        "Kms_Driven",
        "Present_Price",
        "Selling_Price",
        "Depreciation",
        "Depreciation_Pct",
        "Owner"
    ]
].corr()

print(correlation.round(3))

print("\nImportant observations:")
print("Age vs Depreciation %:",
      round(correlation.loc["Age", "Depreciation_Pct"], 3))

print("Kms Driven vs Depreciation %:",
      round(correlation.loc["Kms_Driven", "Depreciation_Pct"], 3))
```

```python
age_selling_corr = data["Age"].corr(
    data["Selling_Price"]
)

kms_selling_corr = data["Kms_Driven"].corr(
    data["Selling_Price"]
)

print("\n Selling Price Relationship")

print("Age vs Selling Price:",
      round(age_selling_corr, 3))

print("Kms Driven vs Selling Price:",
      round(kms_selling_corr, 3))

print(
    "\nConclusion: Age has a weak negative relationship "
    "with selling price, while Kms Driven has almost "
    "no direct linear relationship with selling price."
)
```

```python
two_wheeler_keywords = [
    "Royal Enfield",
    "UM Renegade",
    "KTM",
    "Bajaj Dominar",
    "Bajaj Pulsar",
    "Bajaj Avenger",
    "Bajaj Discover",
    "Bajaj  ct",
    "Hyosung",
    "Mahindra Mojo",
    "Honda CB",
    "Honda CBR",
    "Honda Activa",
    "Honda Dream",
    "Honda Karizma",
    "Honda CB Unicorn",
    "Honda CB Shine",
    "Hero Extreme",
    "Hero Passion",
    "Hero Splender",
    "Hero Honda",
    "Hero Glamour",
    "Hero Hunk",
    "Hero Ignitor",
    "Hero  Ignitor",
    "Hero  CBZ",
    "Yamaha FZ",
    "Yamaha Fazer",
    "TVS Apache",
    "TVS Sport",
    "TVS Jupyter",
    "TVS Wego",
    "Suzuki Access"
]


def is_two_wheeler(name):

    name_lower = name.lower()

    for keyword in two_wheeler_keywords:
        if keyword.lower() in name_lower:
            return True

    return False


data["Vehicle_Type"] = data["Car_Name"].apply(
    lambda x: "Two-wheeler"
    if is_two_wheeler(x)
    else "Car"
)
```

```python
two_wheelers = data[
    data["Vehicle_Type"] == "Two-wheeler"
]

print("\n Two-Wheeler Data")
print("Total Two-wheelers:", len(two_wheelers))

print(
    two_wheelers[
        ["Car_Name", "Year", "Selling_Price",
         "Present_Price", "Kms_Driven"]
    ]
)
```

```python
oldest_bike = two_wheelers.loc[
    two_wheelers["Year"].idxmin()
]

print("\n Oldest Bike")
print(oldest_bike[
    ["Car_Name", "Year", "Selling_Price",
     "Present_Price", "Kms_Driven"]
])
```

```python

newest_bike = two_wheelers.loc[
    two_wheelers["Year"].idxmax()
]

print("\n Newest Bike")
print(newest_bike[
    ["Car_Name", "Year", "Selling_Price",
     "Present_Price", "Kms_Driven"]
])
```

```python
bike_counts = two_wheelers[
    "Car_Name"
].value_counts()

most_sold_bike = bike_counts.idxmax()
most_sold_bike_count = bike_counts.max()

print("\n Most Sold Bike")
print("Bike:", most_sold_bike)
print("Number of Records:", most_sold_bike_count)
```

```python
good_bike_deals = two_wheelers[
    two_wheelers["Retention_Pct"] >= 90
].sort_values(
    "Retention_Pct",
    ascending=False
)

print("\n Two-Wheeler Deals Above General Expectation")

print("Number of good deals:",
      len(good_bike_deals))

print(
    good_bike_deals[
        [
            "Car_Name",
            "Year",
            "Selling_Price",
            "Present_Price",
            "Kms_Driven",
            "Retention_Pct"
        ]
    ].head(10).round(2)
)
```
